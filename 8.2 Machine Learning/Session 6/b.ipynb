**Question:** Implement DBSCAN clustering.

The accompanying CSV file is a small, reproducible input dataset. Run the code cell below to see the implementation and its results.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.metrics import silhouette_score
QID = 27
points = pd.read_csv("b_points.csv")
X = StandardScaler().fit_transform(points[["x", "y"]])
if QID == 26:
    labels = KMeans(n_clusters=3, random_state=69, n_init=10).fit_predict(X)
    print("K-Means cluster sizes:", pd.Series(labels).value_counts().sort_index().to_dict())
elif QID == 27:
    labels = DBSCAN(eps=0.35, min_samples=4).fit_predict(X)
    print("DBSCAN labels (-1 is noise):", pd.Series(labels).value_counts().sort_index().to_dict())
elif QID == 34:
    methods = {
        "Hierarchical": AgglomerativeClustering(n_clusters=3),
        "DBSCAN": DBSCAN(eps=0.35, min_samples=4),
    }
    for name, model in methods.items():
        labels = model.fit_predict(X)
        clusters = set(labels) - {-1}
        score = silhouette_score(X, labels) if len(clusters) > 1 else float("nan")
        print(f"{name}: clusters={len(clusters)}, noise={(labels == -1).sum()}, silhouette={score:.3f}")
else:
    clusters = set(labels) - {-1}
    print("Silhouette:", round(silhouette_score(X, labels), 3) if len(clusters) > 1 else "not defined")
```
