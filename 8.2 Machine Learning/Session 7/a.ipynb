**Question:** Calculate disease probability using Bayes' theorem.

The accompanying CSV file is a small, reproducible input dataset. Run the code cell below to see the implementation and its results.

```python
from pathlib import Path
import numpy as np
import pandas as pd
params = pd.read_csv("a_test_parameters.csv").iloc[0]
prior = float(params.prevalence)
sensitivity = float(params.sensitivity)
false_positive_rate = float(params.false_positive_rate)
true_positive_probability = prior * sensitivity
positive_evidence = true_positive_probability + (1 - prior) * false_positive_rate
posterior = true_positive_probability / positive_evidence
print(f"Prior P(disease): {prior:.3f}")
print(f"Likelihood P(positive | disease): {sensitivity:.3f}")
print(f"Evidence P(positive): {positive_evidence:.3f}")
print(f"Posterior P(disease | positive): {posterior:.3%}")
```
