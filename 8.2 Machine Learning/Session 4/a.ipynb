**Question:** Implement Support Vector Regression.

The accompanying CSV file is a small, reproducible input dataset. Run the code cell below to see the implementation and its results.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
data = pd.read_csv("a_measurements.csv")
X, y = data[["study_hours", "practice_tests"]], data.score
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=69)
model = make_pipeline(StandardScaler(), SVR(kernel='rbf', C=20))
model.fit(X_train, y_train)
pred = model.predict(X_test)
print("MAE:", round(mean_absolute_error(y_test, pred), 3))
print("RMSE:", round(mean_squared_error(y_test, pred) ** 0.5, 3))
print("R^2:", round(r2_score(y_test, pred), 3))
```
