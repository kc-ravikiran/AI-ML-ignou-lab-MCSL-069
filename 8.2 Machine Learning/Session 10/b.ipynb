**Question:** Demonstrate classification, regression, clustering, and association mining.

The accompanying CSV file is a small, reproducible input dataset. Run the code cell below to see the implementation and its results.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from itertools import combinations
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, r2_score
from sklearn.cluster import KMeans
data = pd.read_csv("b_student_records.csv")
X, y = data[["feature_1", "feature_2", "feature_3"]], data.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=69, stratify=y)
classifier = LogisticRegression(max_iter=1000).fit(X_train, y_train)
pred = classifier.predict(X_test)
print("Classification:", {name: round(metric(y_test, pred), 3) for name, metric in
      [("accuracy", accuracy_score), ("precision", precision_score),
       ("recall", recall_score), ("f1", f1_score)]})
regressor = LinearRegression().fit(data[["feature_1", "feature_2"]], data.exam_score)
print("Regression R^2:", round(r2_score(data.exam_score, regressor.predict(data[["feature_1", "feature_2"]])), 3))
labels = KMeans(n_clusters=3, random_state=69, n_init=10).fit_predict(data[["feature_1", "feature_2"]])
print("Clustering group sizes:", pd.Series(labels).value_counts().sort_index().to_dict())
transactions = [set(row.split("|")) for row in pd.read_csv("b_transactions.csv")["items"]]
counts = {}
for transaction in transactions:
    for size in range(1, len(transaction) + 1):
        for itemset in combinations(sorted(transaction), size):
            counts[itemset] = counts.get(itemset, 0) + 1
print("Association-mining itemsets (support >= 0.25):",
      {items: round(count / len(transactions), 2) for items, count in counts.items()
       if count / len(transactions) >= 0.25})
```
