**Question:** Implement Apriori frequent-itemset mining.

The accompanying CSV file is a small, reproducible input dataset. Run the code cell below to see the implementation and its results.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from itertools import combinations
transactions = [set(row.split("|")) for row in pd.read_csv("a_transactions.csv")["items"]]
min_support, min_confidence = 0.25, 0.60
counts = {}
for transaction in transactions:
    for size in range(1, len(transaction) + 1):
        for items in combinations(sorted(transaction), size):
            counts[items] = counts.get(items, 0) + 1
frequent = {items: count for items, count in counts.items()
            if count / len(transactions) >= min_support}
print("Frequent itemsets:")
for items, count in sorted(frequent.items(), key=lambda pair: (len(pair[0]), pair[0])):
    print(items, f"support={count / len(transactions):.2f}")
print("Association rules:")
for items, count in frequent.items():
    if len(items) < 2:
        continue
    for size in range(1, len(items)):
        for antecedent in combinations(items, size):
            consequent = tuple(item for item in items if item not in antecedent)
            confidence = count / frequent[tuple(sorted(antecedent))]
            if confidence >= min_confidence:
                print(f"{antecedent} -> {consequent} (confidence={confidence:.2f})")
```
