**Question:** Solve the 8-puzzle using a search algorithm.

The accompanying CSV file is a small, reproducible input dataset. Run the code cell below to see the implementation and its results.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from heapq import heappop, heappush
row = pd.read_csv("a_puzzle.csv").iloc[0]
start = tuple(map(int, row.initial.split(",")))
goal = tuple(map(int, row.goal.split(",")))
def heuristic(state):
    return sum(abs(i // 3 - goal.index(tile) // 3) + abs(i % 3 - goal.index(tile) % 3)
               for i, tile in enumerate(state) if tile)

queue = [(heuristic(start), 0, start)]
parent = {start: None}
cost = {start: 0}
while queue:
    _, distance, state = heappop(queue)
    if state == goal:
        break
    blank = state.index(0)
    for nxt in [blank - 3, blank + 3, blank - 1, blank + 1]:
        if not 0 <= nxt < 9 or (abs(nxt - blank) == 1 and nxt // 3 != blank // 3):
            continue
        neighbor = list(state)
        neighbor[blank], neighbor[nxt] = neighbor[nxt], neighbor[blank]
        neighbor = tuple(neighbor)
        new_cost = distance + 1
        if new_cost < cost.get(neighbor, float("inf")):
            cost[neighbor] = new_cost
            parent[neighbor] = state
            heappush(queue, (new_cost + heuristic(neighbor), new_cost, neighbor))
path = []
state = goal
while state in parent:
    path.append(state)
    if parent[state] is None:
        break
    state = parent[state]
path.reverse()
print("Moves:", len(path) - 1)
for state in path:
    print(np.array(state).reshape(3, 3))
```
