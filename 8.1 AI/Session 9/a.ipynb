**Question:** Use resolution to classify first-order clauses as satisfiable or unsatisfiable.

The accompanying CSV file is a small, reproducible input dataset. Run the code cell below to see the implementation and its results.

```python
from pathlib import Path
import numpy as np
import pandas as pd
clauses = pd.read_csv("a_clauses.csv").clause.tolist()
# Ground the universally quantified variables over the constants in this finite example.
constants = ["a", "b"]
ground = set()
for clause in clauses:
    for constant in constants:
        ground.add(clause.replace("(x)", f"({constant})").replace("(y)", f"({constant})"))

def parse_clause(text):
    return frozenset(lit.strip() for lit in text.split("|"))

known = {parse_clause(clause) for clause in ground}
changed = True
while changed and frozenset() not in known:
    changed = False
    snapshot = list(known)
    for i, left in enumerate(snapshot):
        for right in snapshot[i + 1:]:
            for literal in left:
                complement = literal[1:] if literal.startswith("~") else "~" + literal
                if complement in right:
                    resolvent = frozenset((left - {literal}) | (right - {complement}))
                    if resolvent not in known:
                        known.add(resolvent)
                        changed = True
                        print("Derived:", " | ".join(sorted(resolvent)) if resolvent else "EMPTY CLAUSE")
print("Result:", "unsatisfiable" if frozenset() in known else "satisfiable (no contradiction derived)")
```
