**Question:** Determine whether a first-order knowledge base entails a query.

The accompanying CSV file is a small, reproducible input dataset. Run the code cell below to see the implementation and its results.

```python
from pathlib import Path
import numpy as np
import pandas as pd
kb = pd.read_csv("a_knowledge_base.csv").fillna("")
facts = set(kb.loc[kb.kind == "fact", "premise"])
rules = [(row.premise, row.conclusion) for _, row in kb[kb.kind == "rule"].iterrows()]
query = kb.loc[kb.kind == "query", "conclusion"].iloc[0]
changed = True
while changed:
    changed = False
    for premise, conclusion in rules:
        if premise in facts and conclusion not in facts:
            facts.add(conclusion)
            changed = True
            print(f"Derived {conclusion} from {premise}")
print(f"KB entails {query}: {query in facts}")
```
