**Question:** Implement Bidirectional Search for a shortest path.

The accompanying CSV file is a small, reproducible input dataset. Run the code cell below to see the implementation and its results.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from collections import deque
edges = pd.read_csv("a_edges.csv")
ends = pd.read_csv("a_endpoints.csv").iloc[0]
graph = {}
for _, edge in edges.iterrows():
    graph.setdefault(edge.source, []).append(edge.target)
    graph.setdefault(edge.target, []).append(edge.source)
start, goal = ends.start, ends.goal
front = {start: None}
back = {goal: None}
meet = None
while front and back and meet is None:
    forward = len(front) <= len(back)
    active, other = (front, back) if forward else (back, front)
    next_layer = {}
    for node in active:
        for neighbor in graph[node]:
            if neighbor not in active and neighbor not in next_layer:
                next_layer[neighbor] = node
            if neighbor in other:
                meet = neighbor
                break
        if meet:
            break
    active.update(next_layer)
    if forward:
        front = active
    else:
        back = active

def reconstruct(parents, node, reverse=True):
    path = []
    while node is not None:
        path.append(node)
        node = parents[node]
    return list(reversed(path)) if reverse else path

if meet is not None:
    left = reconstruct(front, meet)
    right = reconstruct(back, meet, reverse=False)
    path = left + right[1:]
    print("Shortest path:", " -> ".join(path), "| edges:", len(path) - 1)
else:
    print("No path found.")
```
