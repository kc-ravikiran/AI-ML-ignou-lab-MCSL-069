**Question:** Implement Breadth-First Search.

The accompanying CSV file is a small, reproducible input dataset. Run the code cell below to see the implementation and its results.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from collections import deque
edges = pd.read_csv("b_edges.csv")
ends = pd.read_csv("b_endpoints.csv").iloc[0]
graph = {}
for _, edge in edges.iterrows():
    graph.setdefault(edge.source, []).append(edge.target)
    graph.setdefault(edge.target, [])
start, goal = ends.start, ends.goal
queue = deque([start])
parent = {start: None}
while queue and goal not in parent:
    node = queue.popleft()
    for neighbor in graph[node]:
        if neighbor not in parent:
            parent[neighbor] = node
            queue.append(neighbor)
if goal in parent:
    path = []
    node = goal
    while node is not None:
        path.append(node)
        node = parent[node]
    print("Path:", " -> ".join(reversed(path)))
else:
    print("No path found.")
```
