

**Question: a** Write a Python Program to implement Min-Max Algorithm

The accompanying CSV file is a small, reproducible input dataset. Run the code cell below to see the implementation and its results.

```python
# Min-Max Algorithm Implementation

def minimax(depth, node_index, is_max, values, height):
    
    # Base case: leaf node reached
    if depth == height:
        return values[node_index]

    if is_max:
        return max(
            minimax(depth + 1, node_index * 2, False, values, height),
            minimax(depth + 1, node_index * 2 + 1, False, values, height)
        )
    else:
        return min(
            minimax(depth + 1, node_index * 2, True, values, height),
            minimax(depth + 1, node_index * 2 + 1, True, values, height)
        )


# Driver Code
values = [3, 5, 2, 9, 12, 5, 23, 23]

height = 3  # Tree height for 8 leaf nodes (2^3)

result = minimax(0, 0, True, values, height)

print("The optimal value is:", result)
```
