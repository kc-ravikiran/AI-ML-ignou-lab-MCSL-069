**Question:** Implement the AO* algorithm.

The accompanying CSV file is a small, reproducible input dataset. Run the code cell below to see the implementation and its results.

```python
from pathlib import Path
import numpy as np
import pandas as pd
rows = pd.read_csv("a_and_or_graph.csv")
options = {}
for _, row in rows.iterrows():
    children = [] if pd.isna(row.children) else str(row.children).split(",")
    options.setdefault(row.node, []).append((children, float(row.cost)))

def ao_star(node, active=()):
    if node in active:
        raise ValueError("Cycle detected in AND-OR graph")
    choices = options[node]
    scored = [(cost + sum(ao_star(child, active + (node,))[0] for child in children), children)
              for children, cost in choices]
    return min(scored, key=lambda pair: pair[0])

cost, solution = ao_star("S")
print("Optimal cost:", cost)
print("Selected AND-successor group from S:", solution)
```
