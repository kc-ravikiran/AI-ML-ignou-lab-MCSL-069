**Question:** Implement Depth-First Search.

The accompanying CSV file is a small, reproducible input dataset. Run the code cell below to see the implementation and its results.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from collections import deque
edges = pd.read_csv("a_edges.csv")
ends = pd.read_csv("a_endpoints.csv").iloc[0]
graph = {}
for _, edge in edges.iterrows():
    graph.setdefault(edge.source, []).append(edge.target)
    graph.setdefault(edge.target, [])
start, goal = ends.start, ends.goal
stack = [start]
parent = {start: None}
while stack and goal not in parent:
    node = stack.pop()
    for neighbor in reversed(graph[node]):
        if neighbor not in parent:
            parent[neighbor] = node
            stack.append(neighbor)
if goal in parent:
    path = []
    node = goal
    while node is not None:
        path.append(node)
        node = parent[node]
    print("Path:", " -> ".join(reversed(path)))
else:
    print("No path found.")
```
