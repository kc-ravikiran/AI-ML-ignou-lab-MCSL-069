**Question:** Implement Iterative Deepening Depth-First Search.

The accompanying CSV file is a small, reproducible input dataset. Run the code cell below to see the implementation and its results.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from collections import deque
edges = pd.read_csv("b_edges.csv")
ends = pd.read_csv("b_endpoints.csv").iloc[0]
graph = {}
for _, edge in edges.iterrows():
    graph.setdefault(edge.source, []).append(edge.target)
    graph.setdefault(edge.target, [])
start, goal = ends.start, ends.goal
def depth_limited(node, limit, path, seen):
    if node == goal:
        return path
    if limit == 0:
        return None
    for neighbor in graph[node]:
        if neighbor not in seen:
            result = depth_limited(neighbor, limit - 1, path + [neighbor], seen | {neighbor})
            if result:
                return result
    return None

path = None
for depth in range(len(graph) + 1):
    path = depth_limited(start, depth, [start], {start})
    print(f"Depth limit {depth}: {'found' if path else 'not found'}")
    if path:
        break
if path:
    print("Path:", " -> ".join(path))
```
