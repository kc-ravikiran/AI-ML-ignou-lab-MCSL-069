
**Question:** Write a Python Program to Solve N-Queen Problem without using Recursion.

The accompanying CSV file is a small, reproducible input dataset. Run the code cell below to see the implementation and its results.

```python
def solve_n_queens(n):
    board = [-1] * n      # board[row] = column position of queen
    stack = []
    row = 0
    col = 0

    while True:
        while row < n:
            placed = False

            while col < n:
                safe = True

                for r in range(row):
                    c = board[r]

                    # Check same column and diagonals
                    if c == col or abs(c - col) == abs(r - row):
                        safe = False
                        break

                if safe:
                    board[row] = col
                    stack.append((row, col))
                    row += 1
                    col = 0
                    placed = True
                    break

                col += 1

            if not placed:
                if not stack:
                    return None

                row, prev_col = stack.pop()
                board[row] = -1
                col = prev_col + 1

        return board


def print_board(solution):
    n = len(solution)

    for row in range(n):
        for col in range(n):
            if solution[row] == col:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


# Driver Code
n = int(input("Enter value of N: "))

solution = solve_n_queens(n)

if solution:
    print("\nSolution:")
    print_board(solution)
else:
    print("No solution exists.")
```
